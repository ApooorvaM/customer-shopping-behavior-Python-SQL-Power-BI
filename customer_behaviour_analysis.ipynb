```python
import pandas as pd
```

```python
df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe(include = 'all')
```

```python
# Checking the null values
df.isnull().sum()
```

```python
# Replacing null values of each category with their median values
df['Review Rating'] = df.groupby(['Category'])['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# Checking unique values of each column to see if there is any inconsistent formatting issue
for col in df.columns:
    print(df[col].unique())
    print('='*30)

```

```python
df['Frequency of Purchases'] = df['Frequency of Purchases'].replace({'Fortnightly': 'Bi-Weekly','Every 3 Months':'Quarterly'})
```

```python
df['Frequency of Purchases'].unique()
```

```python
# To check duplicate values
df['Customer ID'].value_counts().unique()
```

```python
# To rename column names in snake case
df.columns = df.columns.str.lower().str.replace(' ','_')
df.columns
```

```python
df = df.rename(columns = {'purchase_amount_(usd)' :'purchase_amount'})
```

```python
df.columns
```

```python
# Making a new column Age group to do age group wise analysis later
def age_grp(x):
    if x < 25:
        y = 'young_adult'
    elif 25 <= x < 35:
        y = 'adult'
    elif 35 <= x < 50:
        y = 'middle_aged'
    else:
        y = 'senior'

    return y
```

```python
df['age_group'] = df['age'].apply(age_grp)
```

```python
df[['age','age_group']].head(10)
```

```python
# Mapping purchase frequency in numeric values
df['purchase_freq_in_days'] = df['frequency_of_purchases'].map(
    {
     'Weekly': 7,
     'Annually': 365,
     'Quarterly': 90,
     'Bi-Weekly': 14,
     'Monthly': 30}
)

df['purchase_freq_in_days'].unique()
```

```python
df.head()
```

```python
# Checking if columns 'discount_applied' and 'promo_code_used' are giving exact same information. then one of the column will be redundant

df[df['discount_applied'] != df['promo_code_used']]
```

```python
# Both the columns are exactly same. We can drop promo_code_used

df.drop('promo_code_used', axis = 1)
```

```python
import urllib.parse
from sqlalchemy import create_engine

# Database settings
host = r'localhost\SQLEXPRESS'
database = 'customer_behaviour'

# Build connection string for Windows Authentication
connection_string = (
    f"DRIVER={{ODBC Driver 17 for SQL Server}};"
    f"SERVER={host};"
    f"DATABASE={database};"
    f"Trusted_Connection=yes;"
)

params = urllib.parse.quote_plus(connection_string)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

# Write DataFrame to SQL Server
df.to_sql("customer", engine, if_exists="replace", index=False)
```
